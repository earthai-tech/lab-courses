```python
# Configuration : Colab ou Jupyter local. Exécutez cette cellule en premier.
import os, shutil, urllib.request
DONNEES = ['forages_eau_terrain.csv', 'sondages_miniers_propre.csv']
URL = 'https://raw.githubusercontent.com/earthai-tech/lab-courses/main/inphb/ts/data/'
for f in DONNEES:
    if os.path.exists(f):
        continue
    if os.path.exists(os.path.join('..', 'data', f)):
        shutil.copy(os.path.join('..', 'data', f), f)   # kit hors ligne / dépôt cloné
    else:
        urllib.request.urlretrieve(URL + f, f)            # Colab : téléchargement
    print('OK', f)
```

```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv('sondages_miniers_propre.csv')
eau = pd.read_csv('forages_eau_terrain.csv')
```

## 1. Histogramme des teneurs
À MODIFIER : essayer `bins=10`, `bins=40`, `bins=100`.

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].hist(df['au_gpt'], bins=40, color='orange')
ax[0].set_title('Teneurs brutes'); ax[0].set_xlabel('Au (g/t)')
ax[1].hist(np.log10(df['au_gpt']), bins=40, color='teal')
ax[1].set_title('Teneurs en échelle log10')
plt.show()
```

## 2. Teneurs par lithologie

```python
df.boxplot(column='au_gpt', by='lithologie', figsize=(9, 5))
plt.yscale('log'); plt.axhline(0.5, color='red', ls='--')
plt.ylabel('Au (g/t)'); plt.suptitle(''); plt.show()
```

## 3. Carte des sondages colorée par la teneur moyenne

```python
t = df.groupby('sondage').agg(x=('x_m', 'first'), y=('y_m', 'first'), au=('au_gpt', 'mean'))
plt.figure(figsize=(7, 5))
plt.scatter(t['x'], t['y'], c=np.log10(t['au']), cmap='YlOrRd', s=80, edgecolor='grey')
plt.colorbar(label='log10 teneur moyenne'); plt.xlabel('X (m)'); plt.ylabel('Y (m)')
plt.show()
```

## 4. Forages d’eau : résistivité et débit
Attention : ce fichier contient encore des erreurs (−999). Les retirer avant le graphique.

```python
ok = eau[eau['resistivite_ohm_m'] > 0]
plt.scatter(ok['resistivite_ohm_m'], ok['debit_m3h'], c=ok['sfi'], cmap='viridis')
plt.xscale('log'); plt.colorbar(label='sfi')
plt.axhline(1, color='red', ls='--')
plt.xlabel('Résistivité (ohm.m)'); plt.ylabel('Débit (m3/h)'); plt.show()
```
